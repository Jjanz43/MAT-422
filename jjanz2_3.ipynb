```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42091295)
```

A joint probability distribution describes the probability that two discrete random variables occur simultaneously.

In this example, the rows represent the outcomes of one random variable and the columns represent the outcomes of another random variable. The probabilities must sum to one.

```python
joint = np.array([
    [0.10, 0.20],
    [0.30, 0.40]
])

print("Joint Probability Distribution")
print(joint)

print("\nSum of all probabilities")
print(joint.sum())

row_probs = joint.sum(axis=1)
col_probs = joint.sum(axis=0)

print("\nMarginal Probability of X")
print(row_probs)

print("\nMarginal Probability of Y")
print(col_probs)

plt.figure(figsize=(5,4))

plt.imshow(joint, cmap="Blues")

plt.colorbar(label="Probability")

plt.xticks([0,1],["Y=1","Y=2"])
plt.yticks([0,1],["X=1","X=2"])

plt.title("Joint Probability Distribution")

plt.show()
```

The entries of the table represent the probability of each pair of outcomes occurring together.

Adding every probability gives a total of 1, confirming that the table is a valid joint probability distribution.

The marginal probabilities are obtained by summing across rows or columns. These values describe the probabilities of each variable individually, regardless of the value of the other variable.

correlation measures the strength and direction of a linear relationship between two variables. A positive correlation means both variables tend to increase together, while a negative correlation means one variable decreases as the other increases. Correlation values range from -1 to 1.

```python
x = np.array([1,2,3,4,5,6])

y = np.array([2,3,5,6,8,9])

correlation = np.corrcoef(x,y)[0,1]

print("Correlation Coefficient")

print(correlation)

plt.figure(figsize=(6,4))

plt.scatter(x,y)

plt.xlabel("X")

plt.ylabel("Y")

plt.title("Scatter Plot Showing Positive Correlation")

plt.grid(True)

plt.show()
```

A random sample is a gathered collection of observations that all have an equal chance of being slected to represent the population. As sample size increases. The sample tends to be a better represaentation of the population. Due to the sample approaching the same size as the population with a large enough sample size technically.

```python
population = np.random.normal(50,10,10000)

sample = np.random.choice(population,100)

print("Population Mean")

print(population.mean())

print("\nSample Mean")

print(sample.mean())

print("\nPopulation Standard Deviation")

print(population.std())

print("\nSample Standard Deviation")

print(sample.std())

plt.figure(figsize=(6,4))

plt.hist(sample,bins=12,edgecolor="black")

plt.title("Histogram of Random Sample")

plt.xlabel("Value")

plt.ylabel("Frequency")

plt.grid(True)

plt.show()
```

The sample statistics are close to the population statistics but are not exactly the same because the sample contains only a subset of the population. The histogram displays the distribution of the sampled observations.
As larger random samples are collected, the sample mean tends to become closer to the population mean, illustrating an important idea in statistics.
