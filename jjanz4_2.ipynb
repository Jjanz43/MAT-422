```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1042026)#used todays date at the seed.

sample = np.random.normal(loc=50, scale=5, size=30)

#mle for mean of norm dist.
mean_mle = np.mean(sample)

print("Sample:")
print(sample)

print("MLE Estimate of the Mean:")
print(mean_mle)


#plotting stuff
plt.hist(sample, bins=8, edgecolor= 'black')
plt.axvline(mean_mle, color='red',)
plt.title("Random Sample with MLE Mean")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

The estimated mean is an average of sample values. The red line shows where the MLE estimate is located. Since MLE chooses the parameter that makes the oibserved data most likely, the sample mean is the best estimate of the population mean in this current example.





```python
#Sample data

x= np.array([1,2,3,4,5,6,7])
y= np.array([3,2,5,4,7,6,8])

# line of best fit
m, b = np.polyfit(x, y, 1)
print("Slope:", m)
print("Intercept :",  b)

#predicted values

y_pred = m *x + b

#plotting time

plt.scatter(x,y, label = "Data")
plt.plot (x,y_pred, color= 'blue', label = "Line of Regression")


plt.title("Example For Linear Regression")
plt.xlabel("x)")
plt.ylabel("y")
plt.legend()
plt.show()
```

The regression closely follows the data points, showing positive linear relationship between x and y. The slope indicates that as x increases by one unit, the predicted value of y increases by the approximate value of the slope.
