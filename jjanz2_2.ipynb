```python
#jered janz mat 422

import numpy as np
import matplotlib.pyplot as plt
```

Probability assigns a value between 0 and 1 to every event. The three probability axioms are:

 Every probability is between 0 and 1.
 The probability of the entire sample space equals 1.
If two events cannot occur together, the probability of either event occurring equals the sum of their probabilities.
The following example uses a fair six-sided die to verify these properties.

```python
outcomes = np.arange(1,7)

P = np.ones(6)/6

print("Probabilities")
print(P)

print("\nSum of probabilities")
print(P.sum())

D = {2,4,6}
J = {1}

PD = len(D)/6
PJ = len(J)/6

print("\nP(D)")
print(PD)

print("P(J)")
print(PJ)

print("P(D)+P(J)")
print(PD+PJ)
plt.figure(figsize=(6,3))

plt.bar(outcomes,P)

plt.xlabel("Die Outcome")

plt.ylabel("Probability")

plt.title("Probability Distribution of a Fair Die")

plt.grid(axis="y")

plt.show()
```



The probability values are all between 0 and 1, satisfying the first probability axiom. The probabilities sum to 1, confirming that the entire sample space has probability one.
Events A and B are mutually exclusive because they cannot occur at the same time. Therefore,

P(A\cup B)=P(A)+P(B),
which illustrates the addition rule for disjoint events.



Conditional probability measures the probability that one event occurs given that another event has already occurred.



P(A|B)=\frac{P(A\cap B)}{P(B)}.


The following example uses a deck of cards.

```python
cards = 52

hearts = 13
face_cards = 12
heart_face = 3

P_heart = hearts/cards
P_face = face_cards/cards
P_heart_given_face = heart_face/face_cards

print("P(Heart)")
print(P_heart)

print("P(Face Card)")
print(P_face)

print("P(Heart | Face Card)")
print(P_heart_given_face)
```

```python
labels = ["Heart","Face Card","Heart | Face"]

values = [P_heart,P_face,P_heart_given_face]

plt.figure(figsize=(5,3))

plt.bar(labels,values)

plt.ylabel("Probability")

plt.title("Conditional Probability Example")

plt.show()
```

Knowing that a card is already a face card changes the sample space. Instead of considering all 52 cards, only the 12 face cards remain possible. Among these, three are hearts. Therefore
P(\text{Heart}|\text{Face Card})=\frac{3}{12}=0.25.

This demonstrates that conditional probabilities depend on the information already known.

A discrete random variable takes a countable number of values.

Rolling a die is an example because only the numbers 1 through 6 are possible outcomes.

The expected value represents the long-run average result of many repeated trials.

```python
x = np.arange(1,7)

p = np.ones(6)/6

expected = np.sum(x*p)

variance = np.sum((x-expected)**2*p)

print("Expected Value")
print(expected)

print("\nVariance")
print(variance)
plt.figure(figsize=(6,3))

plt.stem(x,p)

plt.xlabel("Outcome")

plt.ylabel("Probability")

plt.title("Discrete Random Variable")

plt.grid(True)

plt.show()
```



The probability mass function assigns equal probability to each outcome.

The expected value is 3.5, which is the average value expected over many die rolls.

The variance measures how spread out the outcomes are around the expected value.

The stem plot shows that every outcome has the same probability, illustrating a uniform discrete distribution.

```python
x = np.linspace(-4,4,400)

mean = 0
std = 1

pdf = (1/(std*np.sqrt(2*np.pi))) * np.exp(-0.5*((x-mean)/std)**2)

print("Maximum Density")

print(pdf.max())

print("Approximate Total Area")

print(np.trapz(pdf,x))


plt.figure(figsize=(6,4))

plt.plot(x,pdf)

plt.fill_between(x,pdf,alpha=0.3)

plt.title("Standard Normal Distribution")

plt.xlabel("x")

plt.ylabel("Density")

plt.grid(True)

plt.show()
```

he graph represents the probability density function of the standard normal distribution.

Unlike a discrete distribution, the height of the curve is not itself a probability.

Instead, probabilities correspond to the area under the curve.

The numerical integration produces an area very close to 1, confirming that the density function satisfies the requirements of a probability distribution.
