```python
#linear spaces
u = np.array([2,3])
v = np.array([4,5])
c=3

print("vector u =", u)
print("vector v =", v)

print("\n u + v", u + v)
print("c * u",c * u)

zero = np.zeros(4)

print("\nZero vector=", zero)

print("Additive inverse of u= ",-u)



a= -1
b=2
span_vector = a*u + b*v

print("\nExample of a vector in span{u,v}:")
print("2u - v =", span_vector)

print("\nNon-example:")
print("The set {(1,0),(0,1)} is NOT a vector space because")
print("(1,0) + (0,1) = (1,1), which is not contained in the set.")
```

```python
#orthagonal stuff

u = np.array([1.0, 0.0])
v = np.array([0.0,2.0])

dot = np.dot (u,v)

print("Dot Product =", dot)

angle = np.degrees(
    np.arccos(dot/(np.linalg.norm(u)*np.linalg.norm(v)))
)

print("Angle", angle)

projection = (np.dot(u,v)/np.dot(v,v))*v

print("Projection of u onto v =", projection)


w = np.array([1e-12,1])

print("\nRoundoff Example")
print("Dot product =", np.dot(u,w))
```

A vector is orthognal when a vector  after taking the dot product equals zero. This is how you confirm if something is perpendicular. Computing the angle results in a 90 degree angle

```python
plt.figure(figsize=(5,5))

plt.quiver(0,0,u[0],u[1],angles='xy',scale_units='xy',scale=1,label='u')
plt.quiver(0,0,v[0],v[1],angles='xy',scale_units='xy',scale=1,label='v')

plt.xlim(-1,2)
plt.ylim(-1,3)

plt.grid(True)
plt.axis('equal')

plt.title("Orthogonal Vectors")

plt.show()
```

```python

#Gram-schimdt

vectors = [
    np.array([1.,1.,0.]),
    np.array([1.,0.,1.]),
    np.array([0.,1.,1.])
]

Q = []

for vector in vectors:

    q = vector.copy()

    for basis in Q:
        projection = np.dot(q,basis)*basis
        q = q - projection

    q = q/np.linalg.norm(q)

    Q.append(q)

Q = np.column_stack(Q)

print("Orthonormal Basis\n")
print(Q)

print("\nQ^TQ\n")
print(Q.T @ Q)

print("\nIs Q orthonormal?")
print(np.allclose(Q.T @ Q,np.eye(3)))
```

This is for creating an orthogonal basis while maintaing the already present vector space and you have several linear independt vectors.

```python
# Eigenvalues and Eigenvectors


A = np.array([
    [2,1],
    [1,2]
])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A\n")
print(A)

print("\nEigenvalues")
print(eigenvalues)

print("\nEigenvectors")
print(eigenvectors)

print("\nVerification")

for i in range(len(eigenvalues)):

    lam = eigenvalues[i]
    v = eigenvectors[:,i]

    left = A @ v
    right = lam * v

    residual = np.linalg.norm(left-right)

    print(f"\nEigenpair {i+1}")
    print("λ =", lam)
    print("Residual =", residual)
```

Eigenvalues and eigenvectors describe directions that are only scaled by a linear transformation.
